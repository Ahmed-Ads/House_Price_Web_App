## 4. Target Variable Identification

The dataset contains two price-related variables: the total property amount and the price per unit area.

I will investigate their definitions and relationship before selecting the target variable to avoid data leakage.

```python
cols = ['Amount(in rupees)', 'Price (in rupees)', 'Carpet Area']
compare = df[cols].dropna().sample(10, random_state = 42)
compare
```

```python
compare["Carpet Area (numeric)"] = compare["Carpet Area"].str.extract(r"([\d.]+)").astype(float)

compare["Price × Area"] = (
    compare["Price (in rupees)"] * compare["Carpet Area (numeric)"]
)

compare
```
